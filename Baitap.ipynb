Bài 8: Giải hệ phương trình tuyến tính:
ax + by = m
cx + dy = n

```python
a,b,m = map(int,input().split())
c,d,n = map(int,input().split())
D = a*d - b*c             # Định Thức Chính
Dx = b*n - d*m            # Định Thức theo x
Dy = a*n - c*m            # Định Thức theo y
if D != 0 :
    x = Dx/D 
    y = Dy/D
    print(f"Hệ có nghiệm duy nhất x ={x} , y ={y} ")
else:
    if Dx != 0 or Dy != 0 :
        print("Hệ vô nghiệm")
    else :
        print("Hệ có vô số nghiệm")
```

Bài 10: Nhập vào tâm và bán kính của một đường tròn. Sau đó nhập vào một điểm A(x, y) bất kì và kiểm tra xem nó có thuộc đường tròn hay không?

```python
a,b,r = map(int,input().split())
x,y = map(int,input().split())
if (a-x)**2 + (b-y)**2 == r**2 :
    print(f"Điểm A({x},{y}) thuộc đường tròn")
else :
    print("Điểm A({x},{y}) không thuộc đường tròn")
```
