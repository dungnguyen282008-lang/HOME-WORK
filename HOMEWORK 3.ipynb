```python
#14
x = int(input("Nhap so nguyen(nhap -1 de dung)"))
min_value = x
max_value = x
if x == -1 :
    print("Ban chua nhap du lieu nao")
else :
    min_value = x
    max_value = x
while True :
    x = int(input())
    if x == -1 :
        break
    if x < min_value:
        min_value = x
    if x > max_value :
        max_value = x 
print(f"Gia tri be nhat cua chuoi la {min_value}")
print(f"Gia tri lon nhat cua chuoi la {max_value}")
```

```python
#18
n = int(input())
k = n
"""Một số nguyên dương n có dạng $3^k$ ( ví dụ: $1, 3, 9, 27, 81...)
 khi và chỉ khi ta có thể chia liên tiếp n cho 3 cho đến khi thương bằng 1 mà không có bất kỳ phép chia nào bị dư."""
while k % 3 == 0 :
    k //= 3
if  k == 1 :
    print("YES")
else :
    print("NO")

                                       
```

```python
#19
A = int(input("Nhập số nguyên dương A: "))

f1, f2 = 1, 1
while f1 + f2 <= A:
    f1, f2 = f2, f1 + f2
print("Số Fibonacci lớn nhất không vượt quá A là:", f2)

```
